```python
# STEP 2: Check the Python environment

import sys
import os

print("Python version:", sys.version)
print("Current working directory:", os.getcwd())
print("Environment ready!")
%pip install -q ucimlrepo kagglehub

print("Dataset download packages ready!")
# Install the basic packages needed for our project
%pip install -q pandas numpy matplotlib seaborn scikit-learn torch
from ucimlrepo import fetch_ucirepo

# Download UCI Chronic Kidney Disease dataset
ckd_data = fetch_ucirepo(id=336)

# Separate features and target
X_ckd = ckd_data.data.features
y_ckd = ckd_data.data.targets

print("Dataset downloaded successfully!")
print("Number of patient records:", X_ckd.shape[0])
print("Number of input features:", X_ckd.shape[1])

print("\nFirst 5 records:")
display(X_ckd.head())

print("\nTarget column:")
display(y_ckd.head())
import kagglehub
import os

dataset_path = kagglehub.dataset_download(
    "marcklugerhart/chronic-kidney-failure-dataset-reboot"
)

print("Kaggle dataset downloaded!")
print("Dataset folder:", dataset_path)

for root, dirs, files in os.walk(dataset_path):
    for file in files:
        print("File:", os.path.join(root, file))

from ucimlrepo import fetch_ucirepo

# Download Hospital 4 dataset
risk_data = fetch_ucirepo(id=857)

X_risk = risk_data.data.features
y_risk = risk_data.data.targets

print("Hospital 4 dataset downloaded!")
print("Number of records:", X_risk.shape[0])
print("Number of input features:", X_risk.shape[1])

print("\nFeature columns:")
print(X_risk.columns.tolist())

print("\nTarget columns:")
print(y_risk.columns.tolist())

display(X_risk.head())
display(y_risk.head())

print("Hospital 4 — Data type check")

print("\nFeature data types:")
print(X_risk.dtypes)

print("\nTarget values:")
print(y_risk["class"].value_counts(dropna=False))

print("\nMissing values per column:")
print(X_risk.isna().sum())

print("\nUnique values in important columns:")
for col in ["affected", "stage", "grf", "bp limit", "age"]:
    if col in X_risk.columns:
        print(f"\n{col}:")
        print(X_risk[col].value_counts(dropna=False).head(10))

# Hospital 4: remove target leakage columns

X_risk_clean = X_risk.drop(
    columns=["affected", "stage"],
    errors="ignore"
).copy()

y_risk_clean = y_risk["class"].astype(str).str.strip().str.lower()

print("Hospital 4 cleaning check")
print("Original feature count:", X_risk.shape[1])
print("Features after removing leakage columns:", X_risk_clean.shape[1])

print("\nRemoved columns: affected, stage")
print("Target distribution:")
print(y_risk_clean.value_counts())

print("\nRemaining features:")
print(X_risk_clean.columns.tolist())

# Check whether "affected" exactly matches the target
import pandas as pd

check = pd.DataFrame({
    "affected": X_risk["affected"].astype(str).str.strip(),
    "class": y_risk["class"].astype(str).str.strip().str.lower()
})

print("Affected vs target comparison:")
print(pd.crosstab(check["affected"], check["class"]))

print("\nDoes affected perfectly reveal the target?")
print(
    check.groupby("affected")["class"]
    .value_counts()
)

import pandas as pd
import os

# Locate the downloaded Kaggle CSV
csv_path = os.path.join(
    dataset_path,
    "Dataset_insuficiencia_renal_crnica.csv"
)

kaggle_df = pd.read_csv(csv_path)

print("Kaggle dataset shape:", kaggle_df.shape)
print("\nColumn names:")
print(kaggle_df.columns.tolist())

print("\nFirst 5 rows:")
display(kaggle_df.head())

print("\nPossible target columns:")
print([
    c for c in kaggle_df.columns
    if any(word in c.lower() for word in
           ["class", "target", "ckd", "renal", "kidney", "affected"])
])

print("Kaggle dataset shape:", kaggle_df.shape)

target_col = "Insuficiencia renal crónica"

print("\nTarget distribution:")
print(kaggle_df[target_col].value_counts(dropna=False))

print("\nMissing target values:")
print(kaggle_df[target_col].isna().sum())

print("\nDuplicate rows:")
print(kaggle_df.duplicated().sum())

# STEP 10: Clean Kaggle dataset target labels

import pandas as pd
import kagglehub
import os

dataset_path = kagglehub.dataset_download(
    "marcklugerhart/chronic-kidney-failure-dataset-reboot"
)

csv_path = os.path.join(
    dataset_path,
    "Dataset_insuficiencia_renal_crnica.csv"
)

kaggle_df = pd.read_csv(csv_path)

target_col = "Insuficiencia renal crónica"

# Remove records where the target is missing
kaggle_clean = kaggle_df.dropna(subset=[target_col]).copy()

# Normalize target labels
kaggle_clean[target_col] = (
    kaggle_clean[target_col]
    .astype(str)
    .str.strip()
    .str.lower()
)

# Keep only recognized CKD target labels
kaggle_clean = kaggle_clean[
    kaggle_clean[target_col].isin(["si", "no"])
].copy()

print("Original records:", len(kaggle_df))
print("Clean records:", len(kaggle_clean))
print("Removed records:", len(kaggle_df) - len(kaggle_clean))

print("\nTarget distribution:")
print(kaggle_clean[target_col].value_counts())

print("\nMissing target values remaining:",
      kaggle_clean[target_col].isna().sum())


from sklearn.model_selection import train_test_split

kaggle_part_a, kaggle_part_b = train_test_split(
    kaggle_clean,
    test_size=0.5,
    random_state=42,
    stratify=kaggle_clean["Insuficiencia renal crónica"]
)

kaggle_part_a = kaggle_part_a.reset_index(drop=True)
kaggle_part_b = kaggle_part_b.reset_index(drop=True)

print("Hospital 2 — Part A:", kaggle_part_a.shape)
print("Hospital 3 — Part B:", kaggle_part_b.shape)

print("\nPart A target distribution:")
print(kaggle_part_a["Insuficiencia renal crónica"].value_counts())

print("\nPart B target distribution:")
print(kaggle_part_b["Insuficiencia renal crónica"].value_counts())

print("\nTotal records after splitting:",
      len(kaggle_part_a) + len(kaggle_part_b))


# Remove patient ID from both simulated hospital datasets
kaggle_part_a_clean = kaggle_part_a.drop(
    columns=["Identificación del paciente"],
    errors="ignore"
).copy()

kaggle_part_b_clean = kaggle_part_b.drop(
    columns=["Identificación del paciente"],
    errors="ignore"
).copy()

print("Hospital 2 cleaned shape:", kaggle_part_a_clean.shape)
print("Hospital 3 cleaned shape:", kaggle_part_b_clean.shape)

print("\nPatient ID present in Hospital 2:",
      "Identificación del paciente" in kaggle_part_a_clean.columns)

print("Patient ID present in Hospital 3:",
      "Identificación del paciente" in kaggle_part_b_clean.columns)


print("HOSPITAL 2 — PART A")
print(kaggle_part_a_clean.dtypes)
print("\nMissing values:")
print(kaggle_part_a_clean.isnull().sum())

print("\n" + "=" * 50)

print("HOSPITAL 3 — PART B")
print(kaggle_part_b_clean.dtypes)
print("\nMissing values:")
print(kaggle_part_b_clean.isnull().sum())


# Standard English feature names for Kaggle Hospital 2 and 3

column_mapping = {
    "Edad": "age",
    "Presión arterial": "bp",
    "Gravedad específica de la orina": "sg",
    "Albúmina en orina": "al",
    "Azúcar en orina": "su",
    "Glóbulos rojos": "rbc",
    "Células de pus": "pc",
    "Grupos de células de pus": "pcc",
    "Bacterias": "ba",
    "Glucosa en sangre aleatoria": "bgr",
    "Urea en sangre": "bu",
    "Creatinina sérica": "sc",
    "Sodio en sangre": "sod",
    "Potasio en sangre": "pot",
    "Hemoglobina": "hemo",
    "Volumen corpuscular empaquetado": "pcv",
    "Recuento de glóbulos blancos": "wbcc",
    "Recuento de glóbulos rojos": "rbcc",
    "Apetito": "appet",
    "Insuficiencia renal crónica": "class"
}

kaggle_part_a_std = kaggle_part_a_clean.rename(
    columns=column_mapping
).copy()

kaggle_part_b_std = kaggle_part_b_clean.rename(
    columns=column_mapping
).copy()

print("Hospital 2 standardized columns:")
print(kaggle_part_a_std.columns.tolist())

print("\nHospital 3 standardized columns:")
print(kaggle_part_b_std.columns.tolist())

print("\nHospital 2 shape:", kaggle_part_a_std.shape)
print("Hospital 3 shape:", kaggle_part_b_std.shape)

print("\nHospital 2 target values:")
print(kaggle_part_a_std["class"].value_counts())


# Standardize target labels for Kaggle Hospital 2 and 3
# 1 = CKD, 0 = Not CKD

target_mapping = {
    "si": 1,
    "sí": 1,
    "yes": 1,
    "ckd": 1,
    "no": 0,
    "notckd": 0,
    "not ckd": 0
}

for df in [kaggle_part_a_std, kaggle_part_b_std]:
    df["class"] = (
        df["class"]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(target_mapping)
    )

print("Hospital 2 target distribution:")
print(kaggle_part_a_std["class"].value_counts(dropna=False))

print("\nHospital 3 target distribution:")
print(kaggle_part_b_std["class"].value_counts(dropna=False))

print("\nHospital 2 missing target labels:",
      kaggle_part_a_std["class"].isna().sum())

print("Hospital 3 missing target labels:",
      kaggle_part_b_std["class"].isna().sum())


# Standardize UCI dataset target labels
# 1 = CKD, 0 = Not CKD

def standardize_uci_target(target_series):
    labels = (
        target_series.astype(str)
        .str.strip()
        .str.lower()
        .str.replace(r"\s+", "", regex=True)
        .str.replace(r"[\"']", "", regex=True)
    )

    mapping = {
        "ckd": 1,
        "ckd\t": 1,
        "yes": 1,
        "1": 1,
        "notckd": 0,
        "notckd\t": 0,
        "no": 0,
        "0": 0
    }

    return labels.map(mapping)

y_ckd_std = standardize_uci_target(y_ckd["class"])
y_risk_std = standardize_uci_target(y_risk_clean)

print("Hospital 1 — UCI CKD target:")
print(y_ckd_std.value_counts(dropna=False))

print("\nHospital 4 — UCI Risk Factor target:")
print(y_risk_std.value_counts(dropna=False))

print("\nHospital 1 unmapped labels:", y_ckd_std.isna().sum())
print("Hospital 4 unmapped labels:", y_risk_std.isna().sum())


# Check Hospital 1 and Hospital 4 feature datasets

print("Hospital 1 — UCI CKD")
print("Feature shape:", X_ckd.shape)
print("Target shape:", y_ckd_std.shape)
print("\nMissing values in features:",
      X_ckd.isnull().sum().sum())

print("\n" + "=" * 45)

print("Hospital 4 — UCI Risk Factor")
print("Feature shape:", X_risk_clean.shape)
print("Target shape:", y_risk_std.shape)
print("\nMissing values in features:",
      X_risk_clean.isnull().sum().sum())

print("\nHospital 1 feature names:")
print(X_ckd.columns.tolist())

print("\nHospital 4 feature names:")
print(X_risk_clean.columns.tolist())

# Step 18: Standardize Hospital 4 feature names

hospital4_column_mapping = {
    "bp (Diastolic)": "bp",
    "bp limit": "bp_limit"
}

X_risk_std = X_risk_clean.rename(
    columns=hospital4_column_mapping
).copy()

print("Hospital 4 standardized feature names:")
print(X_risk_std.columns.tolist())

print("\nHospital 4 feature shape:", X_risk_std.shape)
print("Hospital 4 target shape:", y_risk_std.shape)

print("\nDuplicate feature names:",
      X_risk_std.columns[X_risk_std.columns.duplicated()].tolist())

# Step 19: Identify common features across hospitals

hospital1_features = set(X_ckd.columns)

hospital2_features = set(kaggle_part_a_std.columns) - {"class"}

hospital3_features = set(kaggle_part_b_std.columns) - {"class"}

hospital4_features = set(X_risk_std.columns)

common_features = sorted(
    hospital1_features
    & hospital2_features
    & hospital3_features
    & hospital4_features
)

print("Hospital 1 feature count:", len(hospital1_features))
print("Hospital 2 feature count:", len(hospital2_features))
print("Hospital 3 feature count:", len(hospital3_features))
print("Hospital 4 feature count:", len(hospital4_features))

print("\nCommon features across all four hospitals:")
print(common_features)

print("\nTotal common features:", len(common_features))

# Step 20: Align the common features across all hospitals

common_features = sorted(
    set(X_ckd.columns)
    & (set(kaggle_part_a_std.columns) - {"class"})
    & (set(kaggle_part_b_std.columns) - {"class"})
    & set(X_risk_std.columns)
)

# Hospital 1
hospital1_X = X_ckd[common_features].copy()
hospital1_y = y_ckd_std.copy()

# Hospital 2
hospital2_X = kaggle_part_a_std[common_features].copy()
hospital2_y = kaggle_part_a_std["class"].copy()

# Hospital 3
hospital3_X = kaggle_part_b_std[common_features].copy()
hospital3_y = kaggle_part_b_std["class"].copy()

# Hospital 4
hospital4_X = X_risk_std[common_features].copy()
hospital4_y = y_risk_std.copy()

# Verify the alignment
for hospital, X, y in [
    ("Hospital 1", hospital1_X, hospital1_y),
    ("Hospital 2", hospital2_X, hospital2_y),
    ("Hospital 3", hospital3_X, hospital3_y),
    ("Hospital 4", hospital4_X, hospital4_y),
]:
    print(f"\n{hospital}")
    print("Features:", X.shape)
    print("Target:", y.shape)
    print("Feature order correct:", X.columns.tolist() == common_features)
    print("Missing target labels:", y.isna().sum())

print("\nCommon feature count:", len(common_features))
print("\nAligned features:", common_features)

# Step 21: Check missing values and data types

hospital_data = {
    "Hospital 1": hospital1_X,
    "Hospital 2": hospital2_X,
    "Hospital 3": hospital3_X,
    "Hospital 4": hospital4_X
}

for name, X in hospital_data.items():
    print("\n" + "=" * 40)
    print(name)
    print("Shape:", X.shape)

    print("\nMissing values:")
    print(X.isnull().sum()[X.isnull().sum() > 0])

    print("\nData types:")
    print(X.dtypes.value_counts())

    print("\nSample values:")
    print(X.head(3).to_string())


import numpy as np
import pandas as pd
import re
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Preserve the same 19-feature order
feature_order = list(common_features)

# Convert range values such as "59 - 66" to their midpoint.
# Handle "< 12", ">= 74", and Excel-style "20-Dec".
def convert_to_numeric(value):
    if pd.isna(value):
        return np.nan

    if isinstance(value, (int, float, np.integer, np.floating)):
        return float(value)

    s = str(value).strip().lower()

    # Correct known Excel date-conversion artifacts in Hospital 4
    if s == "20-dec":
        return 23.5  # midpoint of the intended 20–27 age bin

    s = s.replace("≥", ">=").replace("≤", "<=")

    # Extract numeric values from ranges and inequality bins
    nums = re.findall(r"\d+(?:\.\d+)?", s)

    if not nums:
        return np.nan

    values = [float(n) for n in nums]

    if "-" in s and len(values) >= 2:
        return (values[0] + values[1]) / 2

    if len(values) >= 2:
        return (values[0] + values[1]) / 2

    # Single-value bins: use the boundary as a numeric proxy
    return values[0]


# Convert text categories into numeric codes consistently
category_maps = {
    "appet": {
        "good": 1, "bueno": 1, "normal": 1,
        "poor": 0, "bajo": 0, "malo": 0
    },
    "ba": {
        "notpresent": 0, "ausente": 0,
        "present": 1, "presente": 1
    },
    "pcc": {
        "notpresent": 0, "ausente": 0,
        "present": 1, "presente": 1
    },
    "pc": {
        "normal": 0, "anormal": 1, "abnormal": 1
    },
    "rbc": {
        "normal": 0, "anormal": 1, "abnormal": 1
    }
}

def preprocess_hospital(X, hospital_name):
    X = X[feature_order].copy()

    for col in X.columns:
        if col in category_maps:
            mapped = X[col].astype(str).str.strip().str.lower().map(
                category_maps[col]
            )
            numeric = pd.to_numeric(X[col], errors="coerce")
            X[col] = mapped.fillna(numeric)
        else:
            X[col] = X[col].apply(convert_to_numeric)

    # Impute locally using this hospital's own data only
    imputer = SimpleImputer(strategy="median")
    X_imputed = pd.DataFrame(
        imputer.fit_transform(X),
        columns=feature_order,
        index=X.index
    )

    # Scale locally; no raw data is sent to a central server
    scaler = StandardScaler()
    X_scaled = pd.DataFrame(
        scaler.fit_transform(X_imputed),
        columns=feature_order,
        index=X.index
    )

    print(
        hospital_name,
        "| Shape:", X_scaled.shape,
        "| Missing after imputation:", int(X_scaled.isna().sum().sum()),
        "| Infinite values:", int(np.isinf(X_scaled.to_numpy()).sum())
    )

    return X_scaled.astype(np.float32)


# Run preprocessing separately for each simulated hospital
hospital1_X_processed = preprocess_hospital(hospital1_X, "Hospital 1")
hospital2_X_processed = preprocess_hospital(hospital2_X, "Hospital 2")
hospital3_X_processed = preprocess_hospital(hospital3_X, "Hospital 3")
hospital4_X_processed = preprocess_hospital(hospital4_X, "Hospital 4")

print("\nPreprocessing completed.")
print("Common feature count:", len(feature_order))


import numpy as np
import pandas as pd

# Reproducible experiment
np.random.seed(42)

# Missingness levels for the four simulated hospital clients
missingness_levels = {
    "Hospital 1": 0.05,
    "Hospital 2": 0.15,
    "Hospital 3": 0.25,
    "Hospital 4": 0.40
}

# Keep the preprocessed datasets unchanged
original_hospital_data = {
    "Hospital 1": hospital1_X_processed.copy(),
    "Hospital 2": hospital2_X_processed.copy(),
    "Hospital 3": hospital3_X_processed.copy(),
    "Hospital 4": hospital4_X_processed.copy()
}

# Create separate datasets with simulated missing values
hospital_data_missing = {}

for hospital, X in original_hospital_data.items():
    X_missing = X.copy()
    missing_rate = missingness_levels[hospital]

    # Randomly mask feature cells
    mask = np.random.random(X_missing.shape) < missing_rate
    X_missing = X_missing.mask(mask)

    hospital_data_missing[hospital] = X_missing

    total_cells = X_missing.size
    missing_cells = int(X_missing.isna().sum().sum())
    actual_rate = missing_cells / total_cells * 100

    print(f"\n{hospital}")
    print("Dataset shape:", X_missing.shape)
    print("Target missingness:", f"{missing_rate * 100:.0f}%")
    print("Actual missingness:", f"{actual_rate:.2f}%")
    print("Missing cells:", missing_cells)

print("\nStep 23 completed.")


from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np

hospital_data_imputed = {}

for hospital, X_missing in hospital_data_missing.items():
    # Each simulated hospital imputes using its own data only
    imputer = SimpleImputer(strategy="median")

    X_imputed_array = imputer.fit_transform(X_missing)

    X_imputed = pd.DataFrame(
        X_imputed_array,
        columns=X_missing.columns,
        index=X_missing.index
    )

    hospital_data_imputed[hospital] = X_imputed.astype(np.float32)

    print(f"\n{hospital}")
    print("Shape:", X_imputed.shape)
    print("Missing values remaining:",
          int(X_imputed.isna().sum().sum()))
    print("Infinite values:",
          int(np.isinf(X_imputed.to_numpy()).sum()))

print("\nStep 24 completed: Local imputation finished.")


import torch
import torch.nn as nn
import numpy as np

# Reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Neural network for binary CKD classification
class CKDModel(nn.Module):
    def __init__(self, input_features=19):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_features, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

    def forward(self, x):
        return self.network(x)

# Create a test model
model = CKDModel(input_features=len(common_features)).to(device)

print("Model created successfully!")
print("Input features:", len(common_features))
print("Device:", device)
print("\nModel architecture:")
print(model)
print("\nStep 25 completed.")


# Prepare features and labels for each simulated hospital client

hospital_targets = {
    "Hospital 1": hospital1_y,
    "Hospital 2": hospital2_y,
    "Hospital 3": hospital3_y,
    "Hospital 4": hospital4_y
}

hospital_training_data = {}

for hospital, X in hospital_data_imputed.items():
    y = pd.to_numeric(
        hospital_targets[hospital], errors="coerce"
    )

    # Align feature rows and labels
    valid = y.notna()
    X_ready = X.loc[valid].copy()
    y_ready = y.loc[valid].astype(np.float32)

    hospital_training_data[hospital] = {
        "X": torch.tensor(
            X_ready.to_numpy(dtype=np.float32),
            dtype=torch.float32
        ),
        "y": torch.tensor(
            y_ready.to_numpy(),
            dtype=torch.float32
        ).reshape(-1, 1)
    }

    print(f"\n{hospital}")
    print("X shape:", tuple(hospital_training_data[hospital]["X"].shape))
    print("y shape:", tuple(hospital_training_data[hospital]["y"].shape))
    print("Class counts:")
    print(y_ready.value_counts().sort_index().to_dict())

print("\nStep 26 completed.")


import copy
import torch
import torch.nn as nn

def train_local_model(global_model, client_data, epochs=2, learning_rate=0.001):
    local_model = copy.deepcopy(global_model).to(device)
    local_model.train()

    X = client_data["X"].to(device)
    y = client_data["y"].to(device)

    optimizer = torch.optim.Adam(
        local_model.parameters(),
        lr=learning_rate
    )
    criterion = nn.BCEWithLogitsLoss()

    for epoch in range(epochs):
        optimizer.zero_grad()
        logits = local_model(X)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()

    return copy.deepcopy(local_model.state_dict()), float(loss.item())

print("Local training function created.")
print("Number of simulated clients:", len(hospital_training_data))
print("Step 27A completed.")


import copy
import torch

def fedavg(client_states, client_sizes):
    total_samples = sum(client_sizes)
    global_state = copy.deepcopy(client_states[0])

    for key in global_state:
        global_state[key] = (
            client_states[0][key] * (client_sizes[0] / total_samples)
        )

        for i in range(1, len(client_states)):
            global_state[key] += (
                client_states[i][key]
                * (client_sizes[i] / total_samples)
            )

    return global_state

print("FedAvg aggregation function created.")
print("Aggregation method: Sample-size weighted averaging")
print("Simulated clients:", len(hospital_training_data))
print("Step 27B completed.")


import copy
import numpy as np
import torch

# Start with a fresh global model
global_model = CKDModel(input_features=len(common_features)).to(device)

NUM_ROUNDS = 5
LOCAL_EPOCHS = 2

fedavg_history = []

for round_num in range(1, NUM_ROUNDS + 1):
    client_states = []
    client_sizes = []
    client_losses = []

    for hospital, client_data in hospital_training_data.items():
        state, loss = train_local_model(
            global_model,
            client_data,
            epochs=LOCAL_EPOCHS,
            learning_rate=0.001
        )

        client_states.append(state)
        client_sizes.append(len(client_data["y"]))
        client_losses.append(loss)

    # Aggregate local model parameters
    global_state = fedavg(client_states, client_sizes)
    global_model.load_state_dict(global_state)

    mean_loss = float(np.mean(client_losses))
    fedavg_history.append({
        "round": round_num,
        "mean_local_loss": mean_loss
    })

    print(
        f"Round {round_num}/{NUM_ROUNDS} | "
        f"Mean local training loss: {mean_loss:.4f}"
    )

print("\nBasic Federated Learning completed.")
print("Rounds:", len(fedavg_history))
print("Clients per round:", len(hospital_training_data))

from sklearn.model_selection import train_test_split
import numpy as np
import torch

hospital_train_data = {}
hospital_test_data = {}

for hospital, data in hospital_training_data.items():

    X = data["X"].cpu().numpy()
    y = data["y"].cpu().numpy().ravel()

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    hospital_train_data[hospital] = {
        "X": torch.tensor(X_train, dtype=torch.float32),
        "y": torch.tensor(y_train, dtype=torch.float32).reshape(-1, 1)
    }

    hospital_test_data[hospital] = {
        "X": torch.tensor(X_test, dtype=torch.float32),
        "y": torch.tensor(y_test, dtype=torch.float32).reshape(-1, 1)
    }

    print(f"\n{hospital}")
    print("Training records:", len(X_train))
    print("Testing records:", len(X_test))

print("\nStep 29 completed: Train/test split ready.")

import copy
import numpy as np
import torch
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Start with a fresh global model
global_model = CKDModel(
    input_features=len(common_features)
).to(device)

NUM_ROUNDS = 5
LOCAL_EPOCHS = 2

fedavg_history = []

def evaluate_global_model(model, test_data):
    model.eval()

    all_labels = []
    all_probs = []

    with torch.no_grad():
        for hospital, data in test_data.items():
            X = data["X"].to(device)
            y = data["y"].cpu().numpy().ravel()

            logits = model(X)
            probs = torch.sigmoid(logits).cpu().numpy().ravel()

            all_labels.extend(y.tolist())
            all_probs.extend(probs.tolist())

    y_true = np.array(all_labels).astype(int)
    y_prob = np.array(all_probs)
    y_pred = (y_prob >= 0.5).astype(int)

    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(
            y_true, y_pred, zero_division=0
        ),
        "recall": recall_score(
            y_true, y_pred, zero_division=0
        ),
        "f1": f1_score(
            y_true, y_pred, zero_division=0
        ),
        "roc_auc": roc_auc_score(y_true, y_prob)
    }


for round_num in range(1, NUM_ROUNDS + 1):

    client_states = []
    client_sizes = []
    client_losses = []

    # Each simulated hospital trains using its training records only
    for hospital, client_data in hospital_train_data.items():

        state, loss = train_local_model(
            global_model,
            client_data,
            epochs=LOCAL_EPOCHS,
            learning_rate=0.001
        )

        client_states.append(state)
        client_sizes.append(len(client_data["y"]))
        client_losses.append(loss)

    # Aggregate local model updates using FedAvg
    global_state = fedavg(client_states, client_sizes)
    global_model.load_state_dict(global_state)

    # Evaluate using held-out test records
    metrics = evaluate_global_model(
        global_model,
        hospital_test_data
    )

    result = {
        "round": round_num,
        "mean_local_loss": float(np.mean(client_losses)),
        **metrics
    }

    fedavg_history.append(result)

    print(
        f"Round {round_num}/{NUM_ROUNDS} | "
        f"Loss: {result['mean_local_loss']:.4f} | "
        f"Accuracy: {result['accuracy']:.4f} | "
        f"Precision: {result['precision']:.4f} | "
        f"Recall: {result['recall']:.4f} | "
        f"F1: {result['f1']:.4f} | "
        f"ROC-AUC: {result['roc_auc']:.4f}"
    )

print("\nStep 30 completed: Basic FL evaluation finished.")

import numpy as np
import torch
from sklearn.metrics import confusion_matrix, classification_report

global_model.eval()

all_labels = []
all_probs = []

with torch.no_grad():
    for hospital, data in hospital_test_data.items():
        X = data["X"].to(device)
        y = data["y"].cpu().numpy().ravel()

        logits = global_model(X)
        probs = torch.sigmoid(logits).cpu().numpy().ravel()

        all_labels.extend(y.astype(int).tolist())
        all_probs.extend(probs.tolist())

y_true = np.array(all_labels)
y_prob = np.array(all_probs)
y_pred = (y_prob >= 0.5).astype(int)

print("Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    target_names=["Not CKD (0)", "CKD (1)"],
    zero_division=0
))

print("\nPrediction probability summary:")
print("Minimum:", round(float(y_prob.min()), 4))
print("Maximum:", round(float(y_prob.max()), 4))
print("Mean:", round(float(y_prob.mean()), 4))
print("Predicted Not CKD:", int((y_pred == 0).sum()))
print("Predicted CKD:", int((y_pred == 1).sum()))


from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix
)
import numpy as np
import torch

# 1. Split existing training records into train and validation
fl_train_data = {}
validation_data = {}

for hospital, data in hospital_train_data.items():
    X = data["X"].cpu().numpy()
    y = data["y"].cpu().numpy().ravel()

    X_train, X_val, y_train, y_val = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    fl_train_data[hospital] = {
        "X": torch.tensor(X_train, dtype=torch.float32),
        "y": torch.tensor(y_train, dtype=torch.float32).reshape(-1, 1)
    }

    validation_data[hospital] = {
        "X": torch.tensor(X_val, dtype=torch.float32),
        "y": torch.tensor(y_val, dtype=torch.float32).reshape(-1, 1)
    }

    print(
        hospital,
        "| Train:", len(X_train),
        "| Validation:", len(X_val)
    )

# 2. Train a fresh baseline FL model
global_model = CKDModel(
    input_features=len(common_features)
).to(device)

NUM_ROUNDS = 5
LOCAL_EPOCHS = 5
fedavg_history = []

for round_num in range(1, NUM_ROUNDS + 1):
    client_states = []
    client_sizes = []
    client_losses = []

    for hospital, data in fl_train_data.items():
        state, loss = train_local_model(
            global_model,
            data,
            epochs=LOCAL_EPOCHS,
            learning_rate=0.001
        )
        client_states.append(state)
        client_sizes.append(len(data["y"]))
        client_losses.append(loss)

    global_model.load_state_dict(
        fedavg(client_states, client_sizes)
    )

    fedavg_history.append({
        "round": round_num,
        "mean_local_loss": float(np.mean(client_losses))
    })

    print(
        f"Round {round_num}/{NUM_ROUNDS} | "
        f"Mean local loss: {np.mean(client_losses):.4f}"
    )

# 3. Get validation probabilities
def collect_probabilities(model, datasets):
    model.eval()
    labels, probabilities = [], []

    with torch.no_grad():
        for data in datasets.values():
            X = data["X"].to(device)
            y = data["y"].cpu().numpy().ravel()

            probs = torch.sigmoid(model(X)).cpu().numpy().ravel()

            labels.extend(y.astype(int).tolist())
            probabilities.extend(probs.tolist())

    return np.array(labels), np.array(probabilities)

y_val, val_prob = collect_probabilities(
    global_model, validation_data
)

# 4. Select threshold using validation F1 only
thresholds = np.arange(0.10, 0.91, 0.01)

threshold_results = []
for threshold in thresholds:
    pred = (val_prob >= threshold).astype(int)
    threshold_results.append((
        f1_score(y_val, pred, zero_division=0),
        threshold
    ))

best_val_f1, best_threshold = max(threshold_results)

print("\nSelected threshold:", round(float(best_threshold), 2))
print("Validation F1:", round(float(best_val_f1), 4))

# 5. Evaluate once on the untouched test data
y_test, test_prob = collect_probabilities(
    global_model, hospital_test_data
)
test_pred = (test_prob >= best_threshold).astype(int)

print("\nFinal held-out test results")
print("Accuracy:", round(accuracy_score(y_test, test_pred), 4))
print("Precision:", round(precision_score(
    y_test, test_pred, zero_division=0
), 4))
print("Recall:", round(recall_score(
    y_test, test_pred, zero_division=0
), 4))
print("F1-score:", round(f1_score(
    y_test, test_pred, zero_division=0
), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, test_prob), 4))
print("\nConfusion matrix:")
print(confusion_matrix(y_test, test_pred))

print("\nStep 32 completed.")


import copy
import numpy as np
import pandas as pd
import torch

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

# Experimental settings
NUM_ROUNDS = 5
LOCAL_EPOCHS = 5
CLIP_NORM = 1.0
FIXED_NOISE = 0.15

adaptive_noise = {
    "Hospital 1": 0.05,
    "Hospital 2": 0.10,
    "Hospital 3": 0.20,
    "Hospital 4": 0.30
}

methods = {
    "Normal FL": None,
    "Fixed-DP FL": {
        hospital: FIXED_NOISE for hospital in hospital_train_data
    },
    "Proposed Adaptive-DP FL": adaptive_noise
}

def protect_client_update(global_state, local_state,
                          clip_norm, noise_multiplier):
    # Calculate local model update
    update = {
        key: local_state[key] - global_state[key]
        for key in global_state
    }

    # Global L2 norm of the entire client update
    total_norm = torch.sqrt(sum(
        torch.sum(value.float() ** 2)
        for value in update.values()
    ))

    scale = min(
        1.0,
        clip_norm / (total_norm.item() + 1e-12)
    )

    # Clip update, then add Gaussian noise
    protected_update = {}

    for key, value in update.items():
        clipped = value * scale

        noise = torch.randn_like(clipped) * (
            noise_multiplier * clip_norm
        )

        protected_update[key] = clipped + noise

    return protected_update


def evaluate_probabilities(y_true, probabilities, threshold):
    predictions = (probabilities >= threshold).astype(int)

    return {
        "accuracy": accuracy_score(y_true, predictions),
        "precision": precision_score(
            y_true, predictions, zero_division=0
        ),
        "recall": recall_score(
            y_true, predictions, zero_division=0
        ),
        "f1": f1_score(
            y_true, predictions, zero_division=0
        ),
        "roc_auc": roc_auc_score(y_true, probabilities)
    }


def get_predictions(model, datasets):
    model.eval()
    labels, probabilities = [], []

    with torch.no_grad():
        for data in datasets.values():
            X = data["X"].to(device)
            y = data["y"].cpu().numpy().ravel()

            prob = torch.sigmoid(model(X)).cpu().numpy().ravel()

            labels.extend(y.astype(int).tolist())
            probabilities.extend(prob.tolist())

    return np.array(labels), np.array(probabilities)


all_results = []
all_round_histories = {}
trained_models = {}

for method_name, noise_settings in methods.items():

    print(f"\n{'=' * 15} {method_name} {'=' * 15}")

    # Use the same initial model parameters for each method
    torch.manual_seed(42)
    np.random.seed(42)

    model = CKDModel(
        input_features=len(common_features)
    ).to(device)

    round_history = []

    for round_num in range(1, NUM_ROUNDS + 1):
        global_state = copy.deepcopy(model.state_dict())
        client_updates = []
        client_sizes = []
        client_losses = []

        for hospital, client_data in fl_train_data.items():

            local_state, loss = train_local_model(
                model,
                client_data,
                epochs=LOCAL_EPOCHS,
                learning_rate=0.001
            )

            if noise_settings is None:
                # Normal FL: unprotected local update
                update = {
                    key: local_state[key] - global_state[key]
                    for key in global_state
                }
            else:
                # Fixed or adaptive noise after update clipping
                update = protect_client_update(
                    global_state,
                    local_state,
                    CLIP_NORM,
                    noise_settings[hospital]
                )

            client_updates.append(update)
            client_sizes.append(len(client_data["y"]))
            client_losses.append(loss)

        total_size = sum(client_sizes)

        # Sample-size weighted aggregation of client updates
        new_state = {}

        for key, current_value in global_state.items():
            average_update = sum(
                client_updates[i][key] *
                (client_sizes[i] / total_size)
                for i in range(len(client_updates))
            )

            new_state[key] = current_value + average_update

        model.load_state_dict(new_state)

        y_val, val_prob = get_predictions(model, validation_data)

        # Choose threshold from validation data only
        candidate_thresholds = np.arange(0.10, 0.91, 0.01)

        threshold_scores = [
            f1_score(
                y_val,
                (val_prob >= threshold).astype(int),
                zero_division=0
            )
            for threshold in candidate_thresholds
        ]

        best_index = int(np.argmax(threshold_scores))
        selected_threshold = float(candidate_thresholds[best_index])

        round_metrics = evaluate_probabilities(
            y_val, val_prob, selected_threshold
        )

        round_history.append({
            "round": round_num,
            "validation_f1": round_metrics["f1"],
            "mean_local_loss": float(np.mean(client_losses))
        })

        print(
            f"Round {round_num}/{NUM_ROUNDS} | "
            f"Validation F1: {round_metrics['f1']:.4f} | "
            f"Threshold: {selected_threshold:.2f}"
        )

    # Final evaluation: held-out test set only
    y_val, val_prob = get_predictions(model, validation_data)

    candidate_thresholds = np.arange(0.10, 0.91, 0.01)
    threshold_scores = [
        f1_score(
            y_val,
            (val_prob >= threshold).astype(int),
            zero_division=0
        )
        for threshold in candidate_thresholds
    ]

    selected_threshold = float(
        candidate_thresholds[int(np.argmax(threshold_scores))]
    )

    y_test, test_prob = get_predictions(model, hospital_test_data)

    final_metrics = evaluate_probabilities(
        y_test, test_prob, selected_threshold
    )

    result = {
        "method": method_name,
        "threshold": selected_threshold,
        **final_metrics
    }

    all_results.append(result)
    all_round_histories[method_name] = round_history
    trained_models[method_name] = copy.deepcopy(model.state_dict())

    print("Final test metrics:", final_metrics)
    print("Validation-selected threshold:", round(selected_threshold, 2))

results_df = pd.DataFrame(all_results)

print("\nFINAL TEST COMPARISON")
display(results_df.round(4))

print("\nStep 33 completed.")

import pandas as pd
import numpy as np
import torch

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

hospital_results = []

for result in all_results:
    method_name = result["method"]
    threshold = result["threshold"]

    # Load the saved global model for this method
    model = CKDModel(
        input_features=len(common_features)
    ).to(device)

    model.load_state_dict(trained_models[method_name])
    model.eval()

    for hospital, data in hospital_test_data.items():
        X = data["X"].to(device)
        y_true = data["y"].cpu().numpy().ravel().astype(int)

        with torch.no_grad():
            probabilities = torch.sigmoid(
                model(X)
            ).cpu().numpy().ravel()

        y_pred = (probabilities >= threshold).astype(int)

        hospital_results.append({
            "method": method_name,
            "hospital": hospital,
            "test_records": len(y_true),
            "accuracy": accuracy_score(y_true, y_pred),
            "precision": precision_score(
                y_true, y_pred, zero_division=0
            ),
            "recall": recall_score(
                y_true, y_pred, zero_division=0
            ),
            "f1": f1_score(
                y_true, y_pred, zero_division=0
            ),
            "roc_auc": (
                roc_auc_score(y_true, probabilities)
                if len(np.unique(y_true)) == 2
                else np.nan
            )
        })

hospital_results_df = pd.DataFrame(hospital_results)

print("HOSPITAL-WISE TEST RESULTS")
display(hospital_results_df.round(4))

hospital_results_df.to_csv(
    "hospital_wise_results.csv",
    index=False
)

print("\nSaved: hospital_wise_results.csv")
print("Step 34 completed.")

print("Available trained models:")
print(list(trained_models.keys()))

print("\nTraining clients:")
print(list(fl_train_data.keys()))

print("\nTest clients:")
print(list(hospital_test_data.keys()))

print("\nStep 35 setup check completed.")

import copy
import numpy as np
import pandas as pd
import torch

from sklearn.metrics import accuracy_score, roc_auc_score

def collect_record_losses(model, datasets):
    """Calculate per-record binary cross-entropy losses."""
    model.eval()
    losses = []

    with torch.no_grad():
        for data in datasets.values():
            X = data["X"].to(device)
            y = data["y"].to(device)

            logits = model(X)
            per_record_loss = torch.nn.functional.binary_cross_entropy_with_logits(
                logits, y, reduction="none"
            ).reshape(-1)

            losses.extend(per_record_loss.cpu().numpy().tolist())

    return np.asarray(losses)


mia_results = []

for method_name, saved_state in trained_models.items():

    # Reconstruct the trained global model
    model = CKDModel(
        input_features=len(common_features)
    ).to(device)

    model.load_state_dict(copy.deepcopy(saved_state))

    # Members: records used for local training
    member_losses = collect_record_losses(
        model, fl_train_data
    )

    # Non-members: records not used for model training
    nonmember_losses = collect_record_losses(
        model, validation_data
    )

    # Lower loss is treated as stronger evidence of membership
    attack_scores = np.concatenate([
        -member_losses,
        -nonmember_losses
    ])

    attack_labels = np.concatenate([
        np.ones(len(member_losses), dtype=int),
        np.zeros(len(nonmember_losses), dtype=int)
    ])

    # Exploratory threshold: midpoint of mean member/non-member losses
    threshold_loss = (
        member_losses.mean() + nonmember_losses.mean()
    ) / 2

    # Predict member if record loss is below the threshold
    attack_predictions = (
        attack_scores >= -threshold_loss
    ).astype(int)

    attack_accuracy = accuracy_score(
        attack_labels, attack_predictions
    )

    attack_auc = roc_auc_score(
        attack_labels, attack_scores
    )

    mia_results.append({
        "method": method_name,
        "member_records": len(member_losses),
        "nonmember_records": len(nonmember_losses),
        "mean_member_loss": member_losses.mean(),
        "mean_nonmember_loss": nonmember_losses.mean(),
        "attack_accuracy": attack_accuracy,
        "attack_roc_auc": attack_auc
    })

    print(f"\n========== {method_name} ==========")
    print("Mean member loss:", round(float(member_losses.mean()), 4))
    print("Mean non-member loss:", round(float(nonmember_losses.mean()), 4))
    print("MIA attack accuracy:", round(float(attack_accuracy), 4))
    print("MIA attack ROC-AUC:", round(float(attack_auc), 4))

mia_results_df = pd.DataFrame(mia_results)

print("\nMIA PRIVACY COMPARISON")
display(mia_results_df.round(4))

mia_results_df.to_csv("mia_privacy_results.csv", index=False)
print("\nSaved: mia_privacy_results.csv")
print("Step 36 completed.")


# STEP 37A — Check training history structure

print("Available methods:", list(all_round_histories.keys()))

for method, history in all_round_histories.items():
    print(f"\nMethod: {method}")
    print("History type:", type(history))

    if isinstance(history, list):
        print("Number of rounds:", len(history))
        if len(history) > 0:
            print("First round details:", history[0])
    elif isinstance(history, dict):
        print("Available history keys:", list(history.keys()))
        for key, value in history.items():
            print(f"{key}: {value}")
            break

print("\nFinal results columns:", results_df.columns.tolist())
print("MIA results columns:", mia_results_df.columns.tolist())

print("\nStep 37A completed.")


import os
import matplotlib.pyplot as plt
import pandas as pd

# Create a folder for all project graphs
os.makedirs("project_graphs", exist_ok=True)

# Graph 1: Validation F1 vs Federated Learning Round
plt.figure(figsize=(8, 5))

for method, history in all_round_histories.items():
    rounds = [item["round"] for item in history]
    f1_scores = [item["validation_f1"] for item in history]
    plt.plot(rounds, f1_scores, marker="o", label=method)

plt.xlabel("Federated Learning Round")
plt.ylabel("Validation F1-score")
plt.title("Validation F1-score vs Federated Learning Round")
plt.xticks(range(1, 6))
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("project_graphs/f1_vs_round.png", dpi=300)
plt.show()


# Graph 2: Mean Local Training Loss vs Round
plt.figure(figsize=(8, 5))

for method, history in all_round_histories.items():
    rounds = [item["round"] for item in history]
    losses = [item["mean_local_loss"] for item in history]
    plt.plot(rounds, losses, marker="o", label=method)

plt.xlabel("Federated Learning Round")
plt.ylabel("Mean Local Training Loss")
plt.title("Mean Local Training Loss vs Round")
plt.xticks(range(1, 6))
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("project_graphs/loss_vs_round.png", dpi=300)
plt.show()


# Graph 3: Final model performance comparison
metrics = ["accuracy", "precision", "recall", "f1", "roc_auc"]

results_df.set_index("method")[metrics].plot(
    kind="bar", figsize=(11, 6), ylim=(0, 1)
)

plt.title("Final Performance Comparison of FL Methods")
plt.ylabel("Score")
plt.xlabel("Federated Learning Method")
plt.xticks(rotation=15, ha="right")
plt.legend(title="Metric")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("project_graphs/final_metrics_comparison.png", dpi=300)
plt.show()


# Graph 4: Hospital-wise F1 comparison
hospital_results = pd.read_csv("hospital_wise_results.csv")

hospital_results.pivot(
    index="hospital", columns="method", values="f1"
).plot(kind="bar", figsize=(10, 6), ylim=(0, 1))

plt.title("Hospital-wise F1-score Comparison")
plt.xlabel("Simulated Hospital Client")
plt.ylabel("F1-score")
plt.xticks(rotation=0)
plt.legend(title="Method")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("project_graphs/hospital_wise_f1.png", dpi=300)
plt.show()


# Graph 5: Membership Inference Attack ROC-AUC
plt.figure(figsize=(8, 5))

plt.bar(
    mia_results_df["method"],
    mia_results_df["attack_roc_auc"]
)

plt.axhline(
    y=0.5,
    linestyle="--",
    label="Random-guess reference (AUC = 0.5)"
)

plt.ylabel("MIA ROC-AUC")
plt.title("Membership Inference Attack Comparison")
plt.xticks(rotation=15, ha="right")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("project_graphs/mia_roc_auc.png", dpi=300)
plt.show()


# Confirm saved graph files
print("\nSaved graph files:")
for filename in sorted(os.listdir("project_graphs")):
    print(filename)

print("\nStep 37B completed.")



import os
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs("project_graphs", exist_ok=True)

# Actual simulated missingness and proposed noise multipliers
mapping_df = pd.DataFrame({
    "hospital": [
        "Hospital 1", "Hospital 2",
        "Hospital 3", "Hospital 4"
    ],
    "target_missingness_percent": [5, 15, 25, 40],
    "actual_missingness_percent": [4.82, 14.98, 24.92, 40.87],
    "noise_multiplier": [0.05, 0.10, 0.20, 0.30]
})

# Save the mapping table
mapping_df.to_csv(
    "project_graphs/missingness_noise_mapping.csv",
    index=False
)

# Plot actual missingness against assigned noise
plt.figure(figsize=(8, 5))

plt.plot(
    mapping_df["actual_missingness_percent"],
    mapping_df["noise_multiplier"],
    marker="o"
)

for _, row in mapping_df.iterrows():
    plt.annotate(
        row["hospital"],
        (row["actual_missingness_percent"], row["noise_multiplier"]),
        xytext=(5, 7),
        textcoords="offset points"
    )

plt.xlabel("Actual Simulated Missingness (%)")
plt.ylabel("Assigned Noise Multiplier")
plt.title("Proposed Adaptive Missingness-to-Noise Mapping")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    "project_graphs/missingness_vs_noise.png",
    dpi=300
)
plt.show()

print("Mapping table:")
display(mapping_df)

print("\nSaved: project_graphs/missingness_vs_noise.png")
print("Saved: project_graphs/missingness_noise_mapping.csv")
print("Step 38 completed.")


import os
import pandas as pd

print("========== FINAL MODEL RESULTS ==========")
display(results_df.round(4))

print("\n========== MIA PRIVACY RESULTS ==========")
display(mia_results_df.round(4))

print("\n========== HOSPITAL-WISE RESULTS ==========")
hospital_results = pd.read_csv("hospital_wise_results.csv")
display(hospital_results.round(4))

print("\n========== PROJECT GRAPH FILES ==========")
for filename in sorted(os.listdir("project_graphs")):
    filepath = os.path.join("project_graphs", filename)
    print(f"{filename} — {os.path.getsize(filepath):,} bytes")

print("\n========== TRAINED MODEL CHECK ==========")
for method, state in trained_models.items():
    print(f"{method}: {len(state)} model parameter groups")

print("\nStep 39 completed.")



import os
import json
import torch
import pandas as pd

# Create final project output folders
os.makedirs("ckd_fl_project/models", exist_ok=True)
os.makedirs("ckd_fl_project/results", exist_ok=True)
os.makedirs("ckd_fl_project/graphs", exist_ok=True)

# Save each trained model
for method, state in trained_models.items():
    if method == "Normal FL":
        filename = "normal_fl_model.pth"
    elif method == "Fixed-DP FL":
        filename = "fixed_dp_fl_model.pth"
    else:
        filename = "adaptive_dp_fl_model.pth"

    torch.save(
        state,
        os.path.join("ckd_fl_project/models", filename)
    )
    print("Saved model:", filename)

# Save evaluation results
results_df.to_csv(
    "ckd_fl_project/results/final_model_results.csv",
    index=False
)

mia_results_df.to_csv(
    "ckd_fl_project/results/mia_privacy_results.csv",
    index=False
)

hospital_results.to_csv(
    "ckd_fl_project/results/hospital_wise_results.csv",
    index=False
)

# Save proposed noise mapping
mapping_df.to_csv(
    "ckd_fl_project/results/missingness_noise_mapping.csv",
    index=False
)

# Save feature list and thresholds for backend use
metadata = {
    "project_title": (
        "PRIVACY-PRESERVING FEDERATED LEARNING FRAMEWORK "
        "FOR MULTI-HOSPITAL MEDICAL DATA ANALYSIS"
    ),
    "case_study": "Chronic Kidney Disease prediction",
    "features": list(common_features),
    "adaptive_threshold": float(
        results_df.loc[
            results_df["method"] == "Proposed Adaptive-DP FL",
            "threshold"
        ].iloc[0]
    ),
    "warning": (
        "Research prototype only; not a clinical diagnosis. "
        "Four hospitals are simulated clients from public datasets."
    )
}

with open("ckd_fl_project/results/model_metadata.json", "w") as f:
    json.dump(metadata, f, indent=2)

# Copy all graph files into the project folder
import shutil

for filename in os.listdir("project_graphs"):
    source = os.path.join("project_graphs", filename)
    destination = os.path.join("ckd_fl_project/graphs", filename)
    shutil.copy2(source, destination)

print("\nSaved project files:")
for root, dirs, files in os.walk("ckd_fl_project"):
    for filename in files:
        print(os.path.join(root, filename))

print("\nStep 40A completed.")


# STEP 40B — Check preprocessing variables

print("========== PREPROCESSING OBJECTS ==========")

for name, obj in globals().copy().items():
    if any(word in name.lower() for word in [
        "imputer", "scaler", "encoder", "processor",
        "preprocess", "feature"
    ]):
        if not name.startswith("_"):
            print(name, "->", type(obj).__name__)

print("\n========== FEATURE ORDER ==========")
print(list(common_features))

print("\n========== PROCESSED CLIENT SHAPES ==========")
for hospital, data in hospital_training_data.items():
    print(hospital, "X shape:", tuple(data["X"].shape))

print("\nStep 40B completed.")


# STEP 41A — Inspect preprocessing function and fitted objects

import inspect

print("========== PREPROCESSING FUNCTION ==========")
print(inspect.getsource(preprocess_hospital))

print("\n========== IMPUTER DETAILS ==========")
print("Imputer type:", type(imputer).__name__)

if hasattr(imputer, "statistics_"):
    print("Number of imputation values:", len(imputer.statistics_))
else:
    print("Imputer is not fitted or has no statistics_ attribute.")

print("\n========== SCALER DETAILS ==========")
if "scaler" in globals():
    print("Scaler type:", type(scaler).__name__)
    if hasattr(scaler, "mean_"):
        print("Scaler feature count:", len(scaler.mean_))
    else:
        print("Scaler is not fitted or has no mean_ attribute.")
else:
    print("No variable named scaler found.")

print("\n========== FEATURE ORDER ==========")
print(feature_order)

print("\nStep 41A completed.")


# STEP 41B — Save hospital-specific preprocessing objects

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

os.makedirs("ckd_fl_project/preprocessors", exist_ok=True)

raw_hospital_features = {
    "Hospital 1": hospital1_X,
    "Hospital 2": hospital2_X,
    "Hospital 3": hospital3_X,
    "Hospital 4": hospital4_X,
}

for hospital_name, raw_X in raw_hospital_features.items():
    X = raw_X[feature_order].copy()

    # Reproduce the original feature conversion
    for col in X.columns:
        if col in category_maps:
            mapped = (
                X[col].astype(str).str.strip().str.lower()
                .map(category_maps[col])
            )
            numeric = pd.to_numeric(X[col], errors="coerce")
            X[col] = mapped.fillna(numeric)
        else:
            X[col] = X[col].apply(convert_to_numeric)

    # Fit and save this simulated client's preprocessing
    local_imputer = SimpleImputer(strategy="median")
    X_imputed = local_imputer.fit_transform(X)

    local_scaler = StandardScaler()
    local_scaler.fit(X_imputed)

    safe_name = hospital_name.lower().replace(" ", "_")

    joblib.dump(
        local_imputer,
        f"ckd_fl_project/preprocessors/{safe_name}_imputer.joblib"
    )
    joblib.dump(
        local_scaler,
        f"ckd_fl_project/preprocessors/{safe_name}_scaler.joblib"
    )

    print(
        hospital_name,
        "| Features:", len(feature_order),
        "| Imputer and scaler saved"
    )

print("\nSaved preprocessing files:")
for filename in sorted(
    os.listdir("ckd_fl_project/preprocessors")
):
    print(filename)

print("\nStep 41B completed.")


import os

base = "ckd_fl_project"
os.makedirs(base, exist_ok=True)

html = r"""<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Privacy-Preserving Federated Learning for CKD</title>
  <link rel="stylesheet" href="style.css">
</head>
<body>
<header class="hero">
  <p class="eyebrow">BIOMEDICAL ENGINEERING • RESEARCH PROTOTYPE</p>
  <h1>Privacy-Preserving Federated Learning</h1>
  <p>Multi-Hospital Medical Data Analysis — CKD Case Study</p>
  <span class="notice">Four simulated hospital clients using public datasets</span>
</header>

<nav class="tabs" aria-label="Dashboard sections">
  <button class="tab active" data-page="overview">Overview</button>
  <button class="tab" data-page="hospitals">Hospital Analysis</button>
  <button class="tab" data-page="training">Federated Training</button>
  <button class="tab" data-page="privacy">Privacy Analysis</button>
  <button class="tab" data-page="graphs">Graphs & Images</button>
  <button class="tab" data-page="prediction">CKD Prediction</button>
  <button class="tab" data-page="report">Complete Report</button>
</nav>

<main>
<section id="overview" class="page active">
  <h2>Project Overview</h2>
  <p>This research prototype compares Normal Federated Learning, Fixed-DP FL,
  and a proposed adaptive-noise FL experiment for chronic kidney disease prediction.</p>
  <div class="cards">
    <article class="card"><h3>4</h3><p>Simulated clients</p></article>
    <article class="card"><h3>3</h3><p>FL methods compared</p></article>
    <article class="card"><h3>19</h3><p>Common input features</p></article>
    <article class="card"><h3>5</h3><p>Federated rounds</p></article>
  </div>
  <h3>Client configuration</h3>
  <div id="mappingTable" class="table-wrap"></div>
  <h3>Workflow</h3>
  <div class="workflow">
    <span>Local public data</span><b>→</b><span>Local preprocessing</span>
    <b>→</b><span>Local training</span><b>→</b>
    <span>Clip + noise experiment</span><b>→</b><span>FedAvg server</span>
  </div>
  <p class="muted">The missingness-to-noise mapping is a proposed experimental
  mechanism, not a clinically established rule.</p>
</section>

<section id="hospitals" class="page">
  <h2>Hospital Analysis</h2>
  <p>Performance for each simulated client on its held-out test records.</p>
  <div id="hospitalTable" class="table-wrap"></div>
</section>

<section id="training" class="page">
  <h2>Federated Training</h2>
  <p>Round-wise validation F1 and local training loss, based on recorded runs.</p>
  <div class="chart"><img src="graphs/f1_vs_round.png" alt="Validation F1 by round"></div>
  <div class="chart"><img src="graphs/loss_vs_round.png" alt="Local training loss by round"></div>
  <h3>Final model metrics</h3>
  <div id="metricsTable" class="table-wrap"></div>
</section>

<section id="privacy" class="page">
  <h2>Privacy Analysis</h2>
  <p>Exploratory membership-inference attack results. ROC-AUC near 0.5 means
  this particular attack had limited ability to distinguish member and
  non-member records in this experiment.</p>
  <div class="chart"><img src="graphs/mia_roc_auc.png" alt="MIA ROC-AUC comparison"></div>
  <div id="miaTable" class="table-wrap"></div>
  <p class="warning">This experiment does not establish a formal differential
  privacy guarantee. Noise multipliers alone are not an epsilon/delta guarantee.</p>
</section>

<section id="graphs" class="page">
  <h2>Graphs & Images</h2>
  <div class="gallery">
    <figure><img src="graphs/final_metrics_comparison.png" alt="Final metric comparison"><figcaption>Final metric comparison</figcaption></figure>
    <figure><img src="graphs/hospital_wise_f1.png" alt="Hospital-wise F1"><figcaption>Hospital-wise F1</figcaption></figure>
    <figure><img src="graphs/missingness_vs_noise.png" alt="Missingness and noise"><figcaption>Proposed missingness-to-noise mapping</figcaption></figure>
    <figure><img src="graphs/f1_vs_round.png" alt="F1 by round"><figcaption>Validation F1 by round</figcaption></figure>
    <figure><img src="graphs/loss_vs_round.png" alt="Loss by round"><figcaption>Training loss by round</figcaption></figure>
    <figure><img src="graphs/mia_roc_auc.png" alt="MIA ROC-AUC"><figcaption>MIA comparison</figcaption></figure>
  </div>
</section>

<section id="prediction" class="page">
  <h2>CKD Prediction Prototype</h2>
  <p>Prediction will be enabled only after the Python backend and preprocessing
  compatibility are verified.</p>
  <p class="warning">Not a medical diagnosis. Do not use this prototype for
  treatment decisions.</p>
  <p id="backendStatus">Backend status: not checked.</p>
  <button id="checkBackend" class="primary">Check prediction backend</button>
  <form id="predictionForm">
    <div id="featureInputs" class="form-grid"></div>
    <button type="submit" class="primary">Run research prediction</button>
  </form>
  <div id="predictionResult" aria-live="polite"></div>
</section>

<section id="report" class="page">
  <h2>Complete Report</h2>
  <p>The final report will describe datasets, methodology, evaluation,
  privacy limitations, graphs and conclusions using the actual experiment results.</p>
  <p id="reportStatus">Report PDF will be added in the final packaging step.</p>
  <a class="download" href="results/final_model_results.csv" download>Download final model results CSV</a>
  <a class="download" href="results/hospital_wise_results.csv" download>Download hospital-wise results CSV</a>
  <a class="download" href="results/mia_privacy_results.csv" download>Download MIA results CSV</a>
</section>
</main>

<footer>
  <p>Academic research prototype • Not clinically validated • Not for diagnosis</p>
</footer>
<script src="script.js"></script>
</body>
</html>
"""

css = r"""*{box-sizing:border-box}
:root{font-family:Arial,Helvetica,sans-serif;color:#1e293b;background:#f4f7fb}
body{margin:0}
.hero{background:#102a43;color:white;padding:34px max(5%,calc((100% - 1180px)/2))}
.hero h1{max-width:850px;font-size:clamp(1.8rem,4vw,3rem);margin:10px 0}
.hero p{line-height:1.6}
.eyebrow{letter-spacing:.12em;font-size:.78rem;color:#a9d6e5}
.notice{display:inline-block;border:1px solid #8ebacb;border-radius:20px;padding:8px 12px;margin-top:8px;font-size:.85rem}
.tabs{display:flex;gap:6px;flex-wrap:wrap;background:white;padding:12px max(4%,calc((100% - 1200px)/2));border-bottom:1px solid #d9e2ec}
.tab{border:0;background:#edf2f7;color:#334e68;border-radius:8px;padding:10px 13px;cursor:pointer}
.tab.active{background:#176b87;color:white}
main{max-width:1180px;margin:28px auto;padding:0 18px;min-height:60vh}
.page{display:none}.page.active{display:block}
h2{margin-top:0;color:#102a43}h3{margin-top:26px}
.cards{display:grid;grid-template-columns:repeat(auto-fit,minmax(150px,1fr));gap:14px;margin:22px 0}
.card{background:white;border:1px solid #d9e2ec;border-radius:14px;padding:20px}
.card h3{font-size:2rem;color:#176b87;margin:0}.card p{margin-bottom:0}
.workflow{display:flex;align-items:center;justify-content:center;gap:10px;flex-wrap:wrap}
.workflow span{background:white;border:1px solid #bcccdc;padding:13px;border-radius:10px}
.table-wrap{overflow-x:auto;background:white;border-radius:10px;margin:16px 0}
table{border-collapse:collapse;width:100%;font-size:.9rem}
th,td{padding:10px;border-bottom:1px solid #e2e8f0;text-align:left;white-space:nowrap}
th{background:#eaf2f8}
.chart{background:white;padding:12px;margin:18px 0;border-radius:12px}
.chart img,.gallery img{width:100%;height:auto;display:block}
.gallery{display:grid;grid-template-columns:repeat(auto-fit,minmax(280px,1fr));gap:18px}
figure{margin:0;background:white;border:1px solid #d9e2ec;border-radius:12px;padding:12px}
figcaption{padding:10px 2px;font-weight:bold}
.muted{color:#52667a}.warning{padding:14px;background:#fff4e5;border-left:4px solid #d97706;border-radius:6px;line-height:1.5}
.primary,.download{display:inline-block;border:0;border-radius:8px;background:#176b87;color:white;padding:11px 15px;cursor:pointer;text-decoration:none;margin:8px 8px 8px 0}
.form-grid{display:grid;grid-template-columns:repeat(auto-fit,minmax(210px,1fr));gap:14px;margin:18px 0}
.form-grid label{display:flex;flex-direction:column;gap:6px;font-size:.9rem}
.form-grid input,.form-grid select{width:100%;padding:10px;border:1px solid #bcccdc;border-radius:7px}
#predictionResult{margin:18px 0;padding:14px;background:white;border-radius:10px}
footer{text-align:center;padding:24px;background:#102a43;color:white;margin-top:35px}
@media(max-width:600px){.hero{padding:25px 18px}.tabs{padding:10px}.tab{flex-grow:1}}
"""

js = r"""const tabs=[...document.querySelectorAll(".tab")];
const pages=[...document.querySelectorAll(".page")];
tabs.forEach(btn=>btn.addEventListener("click",()=>{
  tabs.forEach(t=>t.classList.toggle("active",t===btn));
  pages.forEach(p=>p.classList.toggle("active",p.id===btn.dataset.page));
}));

async function loadCSV(path){
  const response=await fetch(path);
  if(!response.ok) throw new Error("Could not load "+path);
  const text=await response.text();
  const rows=text.trim().split(/\r?\n/).map(line=>line.split(","));
  return rows;
}
function renderTable(targetId, rows){
  const target=document.getElementById(targetId);
  if(!rows.length){target.textContent="No results available.";return;}
  const table=document.createElement("table");
  rows.forEach((row,i)=>{
    const tr=document.createElement("tr");
    row.forEach(value=>{
      const cell=document.createElement(i===0?"th":"td");
      cell.textContent=value;
      tr.appendChild(cell);
    });
    table.appendChild(tr);
  });
  target.replaceChildren(table);
}
(async()=>{
  try{
    const mapping=await loadCSV("results/missingness_noise_mapping.csv");
    renderTable("mappingTable",mapping);
    const hospital=await loadCSV("results/hospital_wise_results.csv");
    renderTable("hospitalTable",hospital);
    const metrics=await loadCSV("results/final_model_results.csv");
    renderTable("metricsTable",metrics);
    const mia=await loadCSV("results/mia_privacy_results.csv");
    renderTable("miaTable",mia);
  }catch(error){
    console.error(error);
    document.getElementById("mappingTable").textContent=
      "Open this dashboard through the Python backend to load its data files.";
  }
})();

document.getElementById("checkBackend").addEventListener("click",async()=>{
  const status=document.getElementById("backendStatus");
  try{
    const response=await fetch("/api/health");
    if(!response.ok) throw new Error("Backend unavailable");
    const data=await response.json();
    status.textContent="Backend status: "+data.status;
  }catch(error){
    status.textContent="Backend status: unavailable. Start the Python backend first.";
  }
});

const featureNames=[
  "age","al","appet","ba","bgr","bp","bu","hemo","pc","pcc",
  "pcv","pot","rbc","rbcc","sc","sg","sod","su","wbcc"
];
const numericFeatures=new Set([
  "age","al","bgr","bp","bu","hemo","pcv","pot","rbcc","sc","sg","sod","su","wbcc"
]);
const categoryOptions={
  appet:["good","poor"],ba:["notpresent","present"],
  pc:["normal","abnormal"],pcc:["notpresent","present"],
  rbc:["normal","abnormal"]
};
const inputArea=document.getElementById("featureInputs");
featureNames.forEach(name=>{
  const label=document.createElement("label");
  label.append(document.createTextNode(name.toUpperCase()));
  let input;
  if(categoryOptions[name]){
    input=document.createElement("select");
    const empty=document.createElement("option");
    empty.value="";empty.textContent="Select";input.append(empty);
    categoryOptions[name].forEach(option=>{
      const opt=document.createElement("option");
      opt.value=option;opt.textContent=option;input.append(opt);
    });
  }else{
    input=document.createElement("input");
    input.type="number";input.step="any";
    input.placeholder="Enter value";
  }
  input.name=name;
  input.required=true;
  label.append(input);inputArea.append(label);
});
document.getElementById("predictionForm").addEventListener("submit",async event=>{
  event.preventDefault();
  const result=document.getElementById("predictionResult");
  result.textContent="Sending prediction request...";
  const features={};
  new FormData(event.currentTarget).forEach((value,key)=>features[key]=value);
  try{
    const response=await fetch("/api/predict",{
      method:"POST",headers:{"Content-Type":"application/json"},
      body:JSON.stringify({features})
    });
    const data=await response.json();
    if(!response.ok) throw new Error(data.error||"Prediction failed");
    result.textContent="Research output: "+data.prediction+
      " | Probability: "+Number(data.probability).toFixed(4)+
      " | Not a clinical diagnosis.";
  }catch(error){
    result.textContent="Prediction unavailable: "+error.message+
      ". Backend/model compatibility must be checked first.";
  }
});
"""

for filename, content in [
    ("index.html", html),
    ("style.css", css),
    ("script.js", js),
]:
    path = os.path.join(base, filename)
    with open(path, "w", encoding="utf-8") as f:
        f.write(content)
    print("Created:", path)

print("\nStep 42A completed.")
print("HTML, CSS and JavaScript files are ready.")


from pathlib import Path

project_dir = Path("/content/ckd_fl_project")
project_dir.mkdir(parents=True, exist_ok=True)

app_code = r'''
from pathlib import Path
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import joblib

from flask import Flask, request, jsonify, send_from_directory
from flask_cors import CORS

BASE_DIR = Path(__file__).resolve().parent
DEVICE = torch.device("cpu")

FEATURE_ORDER = [
    "age", "al", "appet", "ba", "bgr", "bp", "bu", "hemo",
    "pc", "pcc", "pcv", "pot", "rbc", "rbcc", "sc", "sg",
    "sod", "su", "wbcc"
]

class CKDModel(nn.Module):
    def __init__(self, input_features=19):
        super().__init__()
        self.network = nn.Sequential(
            nn.Linear(input_features, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1)
        )

    def forward(self, x):
        return self.network(x)

app = Flask(__name__, static_folder=str(BASE_DIR), static_url_path="")
CORS(app)

model = CKDModel(input_features=len(FEATURE_ORDER)).to(DEVICE)
model_path = BASE_DIR / "models" / "adaptive_dp_fl_model.pth"

model_loaded = False
model_message = "Model file not found."

if model_path.exists():
    try:
        state = torch.load(model_path, map_location=DEVICE, weights_only=True)
        model.load_state_dict(state)
        model.eval()
        model_loaded = True
        model_message = "Adaptive-DP model loaded."
    except Exception as exc:
        model_message = f"Could not load model: {exc}"

# Common text encodings for categorical inputs.
# These mappings must be checked against the encodings used during training.
CATEGORY_MAPS = {
    "appet": {"good": 1, "poor": 0},
    "ba": {"present": 1, "notpresent": 0, "no": 0, "yes": 1},
    "pc": {"normal": 1, "abnormal": 0},
    "pcc": {"present": 1, "notpresent": 0, "no": 0, "yes": 1},
    "rbc": {"normal": 1, "abnormal": 0},
}

YES_NO_MAP = {"yes": 1, "no": 0, "true": 1, "false": 0}

def parse_value(feature, value):
    if value is None or str(value).strip() == "":
        return np.nan

    if isinstance(value, (int, float, np.number)):
        return float(value)

    text = str(value).strip().lower()

    if feature in CATEGORY_MAPS and text in CATEGORY_MAPS[feature]:
        return float(CATEGORY_MAPS[feature][text])

    if text in YES_NO_MAP:
        return float(YES_NO_MAP[text])

    # Supports decimal numbers and values such as "120/80"
    try:
        return float(text)
    except ValueError:
        import re
        match = re.search(r"-?\d+(?:\.\d+)?", text)
        if match:
            return float(match.group(0))
        raise ValueError(
            f"Cannot interpret '{value}' for feature '{feature}'. "
            "Use a number or a supported category."
        )

def load_preprocessor(hospital_number):
    folder = BASE_DIR / "preprocessors"
    imputer_path = folder / f"hospital_{hospital_number}_imputer.joblib"
    scaler_path = folder / f"hospital_{hospital_number}_scaler.joblib"

    if not imputer_path.exists() or not scaler_path.exists():
        raise FileNotFoundError(
            f"Preprocessing files for Hospital {hospital_number} were not found."
        )

    return joblib.load(imputer_path), joblib.load(scaler_path)

@app.route("/")
def home():
    return send_from_directory(BASE_DIR, "index.html")

@app.route("/api/health")
def health():
    return jsonify({
        "status": "ok",
        "model_loaded": model_loaded,
        "model_message": model_message,
        "mode": "research demonstration only",
        "feature_count": len(FEATURE_ORDER)
    })

@app.route("/api/predict", methods=["POST"])
def predict():
    if not model_loaded:
        return jsonify({
            "error": "Adaptive-DP model is unavailable.",
            "details": model_message
        }), 503

    payload = request.get_json(silent=True) or {}
    hospital = payload.get("hospital", 1)
    features = payload.get("features", payload.get("data", {}))

    try:
        hospital = int(hospital)
        if hospital not in [1, 2, 3, 4]:
            raise ValueError("Hospital must be 1, 2, 3, or 4.")

        if isinstance(features, list):
            if len(features) != len(FEATURE_ORDER):
                raise ValueError("Provide exactly 19 feature values.")
            row = dict(zip(FEATURE_ORDER, features))
        elif isinstance(features, dict):
            row = features
        else:
            raise ValueError("'features' must be an object or a list.")

        raw = {
            feature: parse_value(feature, row.get(feature))
            for feature in FEATURE_ORDER
        }

        X = pd.DataFrame([raw], columns=FEATURE_ORDER)

        imputer, scaler = load_preprocessor(hospital)
        X_imputed = imputer.transform(X)
        X_scaled = scaler.transform(X_imputed)
        tensor = torch.tensor(X_scaled, dtype=torch.float32, device=DEVICE)

        with torch.no_grad():
            probability = torch.sigmoid(model(tensor)).item()

        # Exploratory display threshold only; not a clinical threshold.
        threshold = 0.60
        predicted_class = int(probability >= threshold)

        return jsonify({
            "prediction": (
                "CKD risk signal" if predicted_class == 1
                else "No CKD risk signal at the selected threshold"
            ),
            "predicted_class": predicted_class,
            "probability": round(float(probability), 4),
            "threshold": threshold,
            "hospital": hospital,
            "features_used": FEATURE_ORDER,
            "warning": (
                "Research demonstration only. Not a medical diagnosis. "
                "Preprocessing compatibility with the original training "
                "pipeline must be verified before interpreting predictions."
            )
        })

    except Exception as exc:
        return jsonify({"error": str(exc)}), 400

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=5000, debug=False)
'''

requirements = """flask
flask-cors
numpy
pandas
scikit-learn
joblib
torch
"""

(project_dir / "app.py").write_text(app_code, encoding="utf-8")
(project_dir / "requirements.txt").write_text(requirements, encoding="utf-8")

print("Created:", project_dir / "app.py")
print("Created:", project_dir / "requirements.txt")
print("Step 42B completed.")



from pathlib import Path
import ast
import joblib
import torch

project_dir = Path("/content/ckd_fl_project")

print("========== BACKEND CHECK ==========")

app_path = project_dir / "app.py"

try:
    ast.parse(app_path.read_text(encoding="utf-8"))
    print("app.py syntax: OK")
except Exception as e:
    print("app.py syntax ERROR:", e)

print("\n========== MODEL CHECK ==========")

model_path = project_dir / "models" / "adaptive_dp_fl_model.pth"

if model_path.exists():
    try:
        state = torch.load(
            model_path,
            map_location="cpu",
            weights_only=True
        )
        print("Adaptive-DP model file: FOUND")
        print("State dictionary entries:", len(state))
    except Exception as e:
        print("Model loading ERROR:", e)
else:
    print("Adaptive-DP model file: NOT FOUND")

print("\n========== PREPROCESSOR CHECK ==========")

for h in range(1, 5):
    imputer_path = (
        project_dir / "preprocessors" /
        f"hospital_{h}_imputer.joblib"
    )
    scaler_path = (
        project_dir / "preprocessors" /
        f"hospital_{h}_scaler.joblib"
    )

    if imputer_path.exists() and scaler_path.exists():
        try:
            imputer = joblib.load(imputer_path)
            scaler = joblib.load(scaler_path)

            print(
                f"Hospital {h}: OK | "
                f"Imputer features={imputer.n_features_in_} | "
                f"Scaler features={scaler.n_features_in_}"
            )
        except Exception as e:
            print(f"Hospital {h}: ERROR - {e}")
    else:
        print(f"Hospital {h}: MISSING preprocessing file(s)")

print("\nStep 42C verification completed.")



from pathlib import Path

app_path = Path("/content/ckd_fl_project/app.py")
code = app_path.read_text(encoding="utf-8")

if "/api/patient-report" in code:
    print("PDF report endpoint already exists. No changes made.")
else:
    addition = r'''

from flask import make_response
from io import BytesIO
from datetime import datetime
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import mm

@app.route("/api/patient-report", methods=["POST"])
def patient_report():
    payload = request.get_json(silent=True) or {}
    patient_id = str(payload.get("patient_id", "Anonymous"))
    hospital = str(payload.get("hospital", "Not specified"))
    features = payload.get("features", {})
    prediction = str(payload.get("prediction", "Not provided"))
    probability = payload.get("probability", None)

    buffer = BytesIO()
    doc = SimpleDocTemplate(
        buffer, pagesize=A4,
        rightMargin=18*mm, leftMargin=18*mm,
        topMargin=18*mm, bottomMargin=18*mm
    )
    styles = getSampleStyleSheet()
    story = [
        Paragraph("CKD Research Prediction Report", styles["Title"]),
        Spacer(1, 8),
        Paragraph("Project: Privacy-Preserving Federated Learning Framework for Multi-Hospital Medical Data Analysis",
                  styles["Normal"]),
        Spacer(1, 8),
        Paragraph("Patient ID: " + patient_id[:80], styles["Normal"]),
        Paragraph("Simulated hospital: " + hospital, styles["Normal"]),
        Paragraph("Report generated: " + datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                  styles["Normal"]),
        Spacer(1, 12),
        Paragraph("Entered patient features", styles["Heading2"])
    ]

    rows = [["Feature", "Value"]]
    if isinstance(features, dict):
        for key, value in features.items():
            rows.append([str(key)[:60], str(value)[:100]])

    table = Table(rows, repeatRows=1, colWidths=[65*mm, 95*mm])
    table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.lightgrey),
        ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("PADDING", (0, 0), (-1, -1), 6)
    ]))
    story += [table, Spacer(1, 12),
              Paragraph("Model result", styles["Heading2"]),
              Paragraph("Result: " + prediction[:200], styles["Normal"])]

    if probability is not None:
        story.append(Paragraph(
            "Model probability: " + str(probability),
            styles["Normal"]
        ))

    story += [
        Spacer(1, 12),
        Paragraph(
            "Research demonstration only. This report is not a medical diagnosis. "
            "Do not use this result alone for clinical decisions. "
            "The model and preprocessing compatibility must be validated before "
            "interpreting predictions.",
            styles["Italic"]
        )
    ]

    doc.build(story)
    response = make_response(buffer.getvalue())
    response.headers["Content-Type"] = "application/pdf"
    response.headers["Content-Disposition"] = 'attachment; filename="ckd_patient_report.pdf"'
    return response
'''
    app_path.write_text(code + addition, encoding="utf-8")
    print("Added PDF report endpoint to app.py")

print("Step 43A completed.")
print("Next: install ReportLab and test PDF generation.")

!pip -q install reportlab flask flask-cors joblib scikit-learn torch pandas numpy

print("Required libraries installed.")



from pathlib import Path

project_dir = Path("/content/ckd_fl_project")
js_path = project_dir / "script.js"

js = js_path.read_text(encoding="utf-8")

old = '''    result.textContent="Research output: "+data.prediction+
      " | Probability: "+Number(data.probability).toFixed(4)+
      " | Not a clinical diagnosis.";'''

new = '''    result.replaceChildren();

    const resultText = document.createElement("p");
    resultText.textContent = "Research output: " + data.prediction +
      " | Probability: " + Number(data.probability).toFixed(4) +
      " | Not a clinical diagnosis.";
    result.appendChild(resultText);

    const pdfButton = document.createElement("button");
    pdfButton.type = "button";
    pdfButton.className = "primary";
    pdfButton.textContent = "Download Patient Report PDF";

    pdfButton.addEventListener("click", async () => {
      pdfButton.disabled = true;
      pdfButton.textContent = "Generating PDF...";

      try {
        const formData = new FormData(
          document.getElementById("predictionForm")
        );
        const featuresForReport = {};
        formData.forEach((value, key) => {
          featuresForReport[key] = value;
        });

        const patientId =
          document.getElementById("patientId")?.value || "Anonymous";
        const hospital =
          document.getElementById("hospitalSelect")?.value || 1;

        await downloadPatientReport(
          patientId,
          hospital,
          featuresForReport,
          data
        );
      } catch (error) {
        alert("PDF download failed: " + error.message);
      } finally {
        pdfButton.disabled = false;
        pdfButton.textContent = "Download Patient Report PDF";
      }
    });

    result.appendChild(pdfButton);'''

if old not in js:
    print("Expected prediction-result code not found.")
    print("No changes made. Paste the latest script.js section for correction.")
else:
    js_path.write_text(js.replace(old, new, 1), encoding="utf-8")
    print("Step 44B code updated.")
    print("Prediction result now creates a PDF download button.")

from pathlib import Path
import ast

project_dir = Path("/content/ckd_fl_project")

# Check Python backend syntax
app_path = project_dir / "app.py"
ast.parse(app_path.read_text(encoding="utf-8"))
print("1. Flask backend syntax: OK")

# Check prediction endpoint and PDF endpoint
app_code = app_path.read_text(encoding="utf-8")

print("2. Prediction endpoint:", "/api/predict" in app_code)
print("3. Patient PDF endpoint:", "/api/patient-report" in app_code)

# Check JavaScript connection
js_code = (project_dir / "script.js").read_text(encoding="utf-8")
print("4. PDF function:", "downloadPatientReport" in js_code)
print("5. PDF button:", "Download Patient Report PDF" in js_code)

# Check model and preprocessors
print(
    "6. Adaptive-DP model:",
    (project_dir / "models/adaptive_dp_fl_model.pth").exists()
)
print(
    "7. Preprocessor files:",
    len(list((project_dir / "preprocessors").glob("*.joblib")))
)

print("\nBackend checks completed.")


%cd /content/ckd_fl_project
!python app.py

import requests

url = "http://127.0.0.1:5000/api/health"

try:
    response = requests.get(url, timeout=10)
    print("HTTP status:", response.status_code)
    print("Response:", response.text)
except Exception as e:
    print("Backend health check failed:", repr(e))
```

```python
%cd /content/ckd_fl_project

from pathlib import Path
import shutil

app_path = Path("app.py")
js_path = Path("script.js")

app = app_path.read_text()
js = js_path.read_text()

# Verify required code exists before changing anything
assert '@app.route("/api/patient-report", methods=["POST"])' in app
assert 'result.appendChild(resultText);' in js
assert 'probability: result.probability ?? result.confidence ?? null' in js

# Backups
shutil.copy2(app_path, "app_backup_before_patient_graph.py")
shutil.copy2(js_path, "script_backup_before_patient_graph.js")

# 1. Add ReportLab graph imports
if "from reportlab.graphics.charts.barcharts import VerticalBarChart" not in app:
    marker = "from reportlab.lib.units import mm"
    assert marker in app
    app = app.replace(
        marker,
        marker + "\nfrom reportlab.graphics.shapes import Drawing\n"
        "from reportlab.graphics.charts.barcharts import VerticalBarChart",
        1
    )

# 2. Read threshold from PDF request
marker = '    probability = payload.get("probability", None)'
if 'threshold = payload.get("threshold", 0.60)' not in app:
    assert marker in app
    app = app.replace(
        marker,
        marker + '\n    threshold = payload.get("threshold", 0.60)',
        1
    )

# 3. Add patient-specific graph to PDF
marker = '''    story += [
        Spacer(1, 12),
        Paragraph(
            "Research demonstration only. This report is not a medical diagnosis. "'''

if "Patient-specific prediction graph" not in app:
    assert marker in app, "PDF disclaimer marker not found"

    graph_code = '''    # Patient-specific probability vs selected threshold chart
    try:
        if probability is not None:
            prob_pct = max(0, min(100, float(probability) * 100))
            threshold_pct = max(0, min(100, float(threshold) * 100))

            drawing = Drawing(165*mm, 78*mm)
            chart = VerticalBarChart()
            chart.x = 18
            chart.y = 14
            chart.width = 130*mm
            chart.height = 52*mm
            chart.data = [[prob_pct, threshold_pct]]
            chart.categoryAxis.categoryNames = [
                "Model probability", "Selected threshold"
            ]
            chart.valueAxis.valueMin = 0
            chart.valueAxis.valueMax = 100
            chart.valueAxis.valueStep = 20
            chart.valueAxis.labelTextFormat = "%d%%"
            drawing.add(chart)

            story += [
                Spacer(1, 12),
                Paragraph(
                    "Patient-specific prediction graph",
                    styles["Heading2"]
                ),
                Paragraph(
                    "This chart compares the model output probability "
                    "with the selected research threshold. It is not "
                    "a clinical diagnostic scale.",
                    styles["Normal"]
                ),
                Spacer(1, 6),
                drawing,
                Spacer(1, 10)
            ]
    except (TypeError, ValueError):
        pass

''' + marker

    app = app.replace(marker, graph_code, 1)

# 4. Add graph under prediction result in dashboard
if "patient-prediction-graph" not in js:
    js_graph = '''    result.appendChild(resultText);

    const probability = Math.max(
      0, Math.min(1, Number(data.probability) || 0)
    );
    const threshold = Math.max(
      0, Math.min(1, Number(data.threshold ?? 0.60))
    );

    const graph = document.createElement("section");
    graph.className = "patient-prediction-graph";
    graph.style.cssText =
      "margin:16px 0;padding:16px;border:1px solid #d5dde5;" +
      "border-radius:12px;max-width:720px;";

    const heading = document.createElement("h3");
    heading.textContent = "Prediction Probability Graph";
    graph.appendChild(heading);

    const note = document.createElement("p");
    note.textContent =
      "Research model output only; not a clinical diagnosis.";
    graph.appendChild(note);

    function addBar(labelText, value, color) {
      const row = document.createElement("div");
      row.style.cssText =
        "display:grid;grid-template-columns:150px minmax(80px,1fr) 58px;" +
        "gap:10px;align-items:center;margin:12px 0;";

      const label = document.createElement("span");
      label.textContent = labelText;

      const track = document.createElement("div");
      track.style.cssText =
        "height:18px;background:#e8edf2;border-radius:5px;overflow:hidden;";

      const bar = document.createElement("div");
      bar.style.cssText =
        "height:100%;width:" + (value * 100) + "%;background:" + color + ";";
      track.appendChild(bar);

      const valueText = document.createElement("strong");
      valueText.textContent = (value * 100).toFixed(1) + "%";

      row.append(label, track, valueText);
      graph.appendChild(row);
    }

    addBar("Model probability", probability, "#0d9488");
    addBar("Selected threshold", threshold, "#64748b");
    result.appendChild(graph);
'''

    js = js.replace("    result.appendChild(resultText);\n", js_graph, 1)

# 5. Pass the threshold to the PDF endpoint
js = js.replace(
    '            probability: result.probability ?? result.confidence ?? null',
    '            probability: result.probability ?? result.confidence ?? null,\n'
    '            threshold: result.threshold ?? 0.60'
)

# Save both files
app_path.write_text(app)
js_path.write_text(js)

# Syntax check
import ast
ast.parse(app)

print("SUCCESS: Dashboard and PDF graph code added.")
print("Backup: app_backup_before_patient_graph.py")
print("Backup: script_backup_before_patient_graph.js")
print("Python syntax: OK")
```
